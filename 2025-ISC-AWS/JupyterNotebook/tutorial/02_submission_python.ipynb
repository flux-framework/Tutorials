#### Chapter 2: **All the Different Ways to Do Work (from _inside_ Python)**

I think this is pretty neat stuff. From inside a Python interpreter, we can get a "handle," basically a connection to our Flux instance, for doing work (like submitting jobs!)

You'll need the `flux` package from Python to do any of this. You can get that through `pip install flux-python`, or by using the python against which your `flux` binary was compiled, accessible through `flux python`. 

````python
import flux
handle = flux.Flux()
````

#### **The jobspec**

In Flux, the basic unit of a job is the jobspec. In python, a jobspec is represented as a `Dict` (since flux-core v0.74.0). We offer a convenience wrapper `.from_command` to create a minimum jobspec from a single command. With constructing a bunch of jobspecs, we can use a handle to submit all of them.

````python
import flux.job
jobspec = flux.job.JobspecV1.from_command(["sleep", "20"])
flux.job.submit(handle, jobspec)
````
